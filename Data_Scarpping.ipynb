## **Import Library**

```python
# Import Library
from google_play_scraper import app, reviews_all, reviews, Sort
import pandas as pd
```

## **Scraping review aplikasi Access by KAI**

```python
# Scraping 50000 reviews
scrapreview = reviews(
    'com.kai.kaiticketing',
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=55000
)
```

```python
scrapreview[0]
```

```python
# Menyimpan hasil data scraping ke file csv
data = pd.DataFrame(scrapreview[0])
data = data[['reviewId','content','score']]
data.to_csv('ulasan_aplikasi.csv', index=False)
print("Data scraping selesai, data sudah disimpan ke file ulasan_aplikasi.csv")
```

Data review aplikasi KAI Access disimpan ke dalam file format CSV. file `ulasan_aplikasi.csv` yang akan digunakan untuk analisis lanjutan yaitu ekstraksi fitur dan pelabelan data, pelatihan menggunakan algoritma ML dan DL untuk analisis sentimen.
